# Le slicing (arr[1:3])
À quoi ça sert : extraire une portion précise et connue d'un tableau, en fonction de sa position.
Tu l'utilises quand tu sais où se trouve ce que tu veux, par exemple :

Prendre les 10 premières lignes d'un dataset pour un aperçu rapide → data[0:10]
Séparer un jeu de données en train/test selon l'ordre → data[:800] (train) et data[800:] (test)
Extraire une période temporelle précise dans une série chronologique (ex. les 30 derniers jours) → data[-30:]

```python
arr[1:3]
```

 prend tous les éléments à partir de l'indice 3 jusqu'à la fin du tableau

```python
arr2[3:]
```

arr2[:3] prend tous les éléments depuis le début jusqu'à l'indice 3 (exclu)

```python
arr2[:3]
```
